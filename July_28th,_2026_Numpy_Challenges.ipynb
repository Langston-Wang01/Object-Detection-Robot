```python
# Challenge 1
# Create a 2D array of random integers of shape (4, 5). Find the sum, mean, min, max, and standard deviation of the whole array.
# Then find the sum of each row and the sum of each column separately using the axis parameter. Print everything clearly labeled.
```

```python
import numpy as np

twoDArray = np.random.randint(0,255, size = (4,5))
print("2D Array: " + "\n" + str(twoDArray))
print("Data Type: " + str(twoDArray.dtype))

Sum = np.sum(twoDArray)
Mean = np.mean(twoDArray)
Min = np.min(twoDArray)
Max = np.max(twoDArray)
Std = np.std(twoDArray)


print("Sum: " + str(Sum) + "\n" "Mean " + str(Mean) + "\n" + "Min: " + str(Min) + "\n" "Max " + str(Max) + "\n" + "Standard Deviation: " + str(Std))

# Axis Sums
columnSum = np.sum(twoDArray, axis = 0) # axis = 0 is column calculation | axis = 1 is row calculation
rowSum = np.sum(twoDArray, axis = 1) #Prints out a List of sums
i = 0
j = 0

for element in rowSum:
  print(f"Column {i}: " + str(element))
  i += 1
print("\n")
for element in columnSum:
  print(f"Row {j}: " + str(element))
  j += 1


```

```python
# Challenge 2 Create a 1D array of 20 random integers between 0 and 100. Write code that returns only the values greater than 50 without using a loop.
# Then replace all values less than 30 with zero. Print the array before and after.
```

```python
array = np.random.randint(0, 100, size = (1, 20))
print("Before: " + "\n" +str(array) +"\n")
# Returns Values greater than 20

greaterValues = array[array > 50]

print("Values Above 50: " + str(greaterValues) +"\n")

array[array < 30] = 0
print("After: " + "\n" + str(array))
```

```python
# Challenge 3 Create two 2D arrays of shape (3, 4) and stack them both vertically and horizontally.
# Then take a (4, 6) array and split it into three equal parts along the column axis. Print the matrix of every result.
```

```python
arrayOne = np.ones((3,4))
arrayTwo = np.zeros((3,4))

hStack = np.hstack([arrayOne, arrayTwo])
vStack = np.vstack([arrayOne, arrayTwo])

print(str(hStack) + "\n")
print(str(vStack) + "\n")

arrayThree = np.random.randint(1,5, size = (4,6)) # Split by 2 each time
print("Original Array: " + "\n" + str(arrayThree) + "\n")
exclude = 2 ## :2 | 2:4 | 4:
include = 0

while exclude <= 6:
  arraySplit = arrayThree[:, include:exclude]
  print(str(arraySplit) + "\n")
  include += 2
  exclude += 2
```

```python
# Challenge 4 Create a (5, 5) array of random floats between 0 and 1.
# Normalize(Min-Max) it so every value falls between 0 and 255 as integers.
```

```python
floatArray = np.random.random((5,5))
min = np.min(floatArray)
max = np.max(floatArray)
difference = (max - min)

normalizedArray = ((floatArray - min) / difference) * 255

print(normalizedArray)



```
