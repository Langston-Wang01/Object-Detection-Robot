```python
""" Challenge #3 Rotate an Image 30 degrees """
```

```python
image = cv2.imread("baby-duck.jpg")
print(image.shape) # Center Coordinate is [190, 320]

rotated_image = np.zeros([380, 640, 3])

pi = np.pi
cv2_imshow(image)

for i in range(640): # Nested For loops to find coordinate relative to center to utilize rotation formula
  shifted_x_coord = i - 320
  for j in range(380):
    shifted_y_coord = j - 190
    new_x = int(shifted_x_coord * np.cos(pi/6) - shifted_y_coord * np.sin(pi/6)) + 320
    new_y = int(shifted_x_coord * np.sin(pi/6) + shifted_y_coord * np.cos(pi/6)) + 190

    if not(new_x < 0 or new_y < 0 or new_y > 379 or new_x > 639):
      rotated_image[new_y, new_x, :] = image[j, i, :]

cv2_imshow(rotated_image)
```
